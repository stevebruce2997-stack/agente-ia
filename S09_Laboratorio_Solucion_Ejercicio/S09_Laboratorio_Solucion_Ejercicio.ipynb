```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

## Ejemplo 1: alquiler de scooters
Queremos estimar **cuánto pagará un cliente que usa un scooter 45 minutos**.

### Paso 1: la persona escribe las reglas

```python
def pago_con_regla(minutos):
    if minutos <= 0:               # sin minutos no hay pago
        return 0
    pago = 0
    for minuto in range(minutos):  # cada minuto suma 0.60
        pago = pago + 0.60
    return round(pago, 2)


print("Predicción para 45 min: S/.", pago_con_regla(45))
```

### Paso 2: el modelo aprende la tarifa

```python
minutos = [[5], [10], [15], [20], [25], [30], [35], [40]]  # entrada (filas)
pagos = [3, 4.5, 7.5, 9, 12.5, 14.5, 18, 20.5]             # respuestas

modelo = LinearRegression(fit_intercept=False)  # parte del cero
modelo.fit(minutos, pagos)                      # aprende

tarifa = modelo.coef_[0]                        # la tarifa
prediccion = modelo.predict([[45]])             # predice 45 min
print("Tarifa:", round(tarifa, 2))
print("Predicción para 45 min: S/.", round(prediccion[0], 2))

plt.scatter(minutos, pagos, color="black")
plt.plot([0, 50], [0, 50 * tarifa], color="blue")
plt.xlabel("Minutos de uso")
plt.ylabel("Pago (S/.)")
plt.show()
```

### Paso 3: un dato mal digitado
Agregamos un error: un alquiler de 30 minutos con S/. 45 en lugar de S/. 15.

```python
minutos_error = minutos + [[30]]
pagos_error = pagos + [45]                      # 30 min con S/. 45

modelo_malo = LinearRegression(fit_intercept=False)
modelo_malo.fit(minutos_error, pagos_error)

tarifa_mala = modelo_malo.coef_[0]
prediccion_mala = modelo_malo.predict([[45]])
print("Tarifa:", round(tarifa_mala, 2))
print("Predicción para 45 min: S/.", round(prediccion_mala[0], 2))

plt.scatter(minutos, pagos, color="black")
plt.scatter([30], [45], color="red")
plt.plot([0, 50], [0, 50 * tarifa], color="gray", linestyle="--")
plt.plot([0, 50], [0, 50 * tarifa_mala], color="red")
plt.xlabel("Minutos de uso")
plt.ylabel("Pago (S/.)")
plt.show()
```

## Ejemplo 2: costo de la matrícula (datos ficticios)
Queremos estimar **cuánto costará matricularse en 15 créditos**.

### Paso 1: la persona escribe las reglas

```python
def costo_con_regla(creditos):
    if creditos <= 0:                # sin créditos no hay costo
        return 0
    costo = 0
    for credito in range(creditos):  # cada crédito suma 560
        costo = costo + 560
    return costo


print("Predicción para 15 créditos: S/.", costo_con_regla(15))
```

### Paso 2: el modelo aprende el costo por crédito

```python
creditos = [[12], [14], [16], [18], [20], [22]]
costos = [7300, 8400, 9700, 10800, 12100, 13200]

modelo2 = LinearRegression(fit_intercept=False)  # parte del cero
modelo2.fit(creditos, costos)                    # aprende

costo_credito = modelo2.coef_[0]                 # costo por crédito
prediccion2 = modelo2.predict([[15]])            # predice 15 créditos
print("Costo por crédito: S/.", round(costo_credito, 2))
print("Predicción para 15 créditos: S/.", round(prediccion2[0], 2))

plt.scatter(creditos, costos, color="black")
plt.plot([0, 25], [0, 25 * costo_credito], color="blue")
plt.xlabel("Créditos matriculados")
plt.ylabel("Costo de la matrícula (S/.)")
plt.show()
```
